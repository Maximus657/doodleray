## Takeaways

- Literal 100 ₽ in the current Doodle balance equals about 28.9 free days, versus 6.7 visible device-days at EJX.
- On July's 57 paid-friend baseline, literal copying would create about 55 service-months before any incremental lift.
- A 100 ₽ renewal discount preserves a payment event and creates zero free service-months.
- With a 5,000 ₽ hard budget and $5.34 incremental contribution per new paid referral, conservative break-even is about 12 incremental payers at the absolute cap; with the expected 3,000–4,000 ₽ issued liability, it is about 8–10.
- The Mini App should initially be one referral page over the existing backend, not a bot migration or game.
